```python
from runze.bus import RunzeBus
from runze.sy01b import SY01B
from runze.sv07m import SV07M
```

```python
bus = RunzeBus(port="COM5", baudrate=9600)
```

```python
pump = SY01B(bus, address=0x01)
```

```python
pump.query_rs485_baud_rate()
pump.query_auto_reset()
pump.query_valve_state()
pump.query_motor_status()
pump.query_valve_status()
pump.query_syringe_position()
```

```python
selector = SV07M(bus, address=0x00)
```

```python
selector.query_address()
```

```python
selector.query_rs485_baudrate()
selector.query_auto_reset()
selector.query_current_port()
selector.query_motor_status()
```

```python
import time

selector.move_to_port(1)
selector.query_motor_status()

time.sleep(1)

selector.query_motor_status()
selector.query_current_port()
```

```python
selector.query_current_port()
```

```python
import time

selector.reset_to_home()
time.sleep(0.1)
selector.query_motor_status()
```

```python
selector.query_current_port()
```

```python
selector.emergency_stop()
```
