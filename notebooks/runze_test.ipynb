```python
with RunzeBus(port="COM5", trace=True) as bus:
    selector = SV07M(bus, address=0x00)

    start_port = selector.query_current_port()
    print("Initial SV07M port:", start_port)

    if start_port != 1:
        raise RuntimeError(
            f"Expected SV07M to start at port 1, got {start_port}."
        )

    selector.move_to_port(2, wait=True)

    port = selector.query_current_port()
    print("SV07M port after move to 2:", port)

    if port != 2:
        raise RuntimeError(f"Expected SV07M port 2, got {port}.")

    selector.move_to_port(1, wait=True)

    port = selector.query_current_port()
    print("SV07M port after move back to 1:", port)

    if port != 1:
        raise RuntimeError(f"Expected SV07M port 1, got {port}.")
```

```python
with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)
    selector = SV07M(bus, address=0x00)

    initial_position = syringe.query_syringe_position()
    motor_status = syringe.query_motor_status()
    valve_state = syringe.query_valve_state()
    selector_port = selector.query_current_port()

    print("Initial position:", initial_position)
    print("Motor:", motor_status)
    print("Valve state:", valve_state)
    print("SV07M port:", selector_port)

    if motor_status != 0:
        raise RuntimeError(f"Expected motor idle, got {motor_status}")

    if valve_state != 1:
        raise RuntimeError(f"Expected SY01B valve state 1, got {valve_state}")

    if selector_port != 6:
        raise RuntimeError(f"Expected SV07M port 6, got {selector_port}")

    volume_ml = 0.2
    rpm = 10

    step_delta = syringe.steps_from_volume_ml(volume_ml)
    current_position = syringe.query_syringe_position()
    target_position = current_position - step_delta

    syringe._validate_syringe_position(target_position)

    syringe.set_dynamic_speed_rpm(rpm)
    syringe.move_plunger_to_position(target_position, wait=True)

    print("Final position:", syringe.query_syringe_position())
```
