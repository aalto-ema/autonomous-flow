```python
from runze.bus import RunzeBus
from runze.sy01b import SY01B

import time
```

```python
with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    print("SY01B valve state:", syringe.query_valve_state())
    print("SY01B valve status:", syringe.query_valve_status())
    print("SY01B motor status:", syringe.query_motor_status())
    print("SY01B reported syringe position:", syringe.query_syringe_position())
```

```python
with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    syringe.set_reset_stall_current(6)
```

```python
with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    start_position = syringe.query_syringe_position()
    print("Initial position:", start_position)

    if start_position != 0:
        raise RuntimeError(
            f"Expected position 0 before this test, got {start_position}."
        )

    syringe.synchronize_syringe_position()
    syringe.set_dynamic_speed_rpm(10)

    started = time.monotonic()
    syringe.move_plunger_to_position(500, wait=True)
    elapsed = time.monotonic() - started

    final_position = syringe.query_syringe_position()
    print("Position after first move:", final_position)
    print(f"First command-to-confirmation time: {elapsed:.2f} s")

    if final_position != 500:
        raise RuntimeError(
            f"Expected position 500 after first move, got {final_position}."
        )
```

```python
# Synchronizing ready from above

with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    syringe.set_dynamic_speed_rpm(10)

    started = time.monotonic()
    syringe.move_plunger_to_position(500, wait=True)
    elapsed = time.monotonic() - started

    position = syringe.query_syringe_position()
    print("Position after move:", position)
    print(f"Command-to-confirmation time: {elapsed:.2f} s")
```

```python
# No power cycle after previous move

with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    start_position = syringe.query_syringe_position()
    print("Initial position:", start_position)

    if start_position != 500:
        raise RuntimeError(
            f"Expected position 500 before this test, got {start_position}."
        )

    # No synchronization.
    syringe.set_dynamic_speed_rpm(10)

    started = time.monotonic()
    syringe.move_plunger_to_position(1000, wait=True)
    elapsed = time.monotonic() - started

    final_position = syringe.query_syringe_position()
    print("Position after move:", final_position)
    print(f"Command-to-confirmation time: {elapsed:.2f} s")

    if final_position != 1000:
        raise RuntimeError(
            f"Expected position 1000 after move, got {final_position}."
        )
```

```python
# After power cycle

with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    print("SY01B motor status:", syringe.query_motor_status())
    print("SY01B reported syringe position:", syringe.query_syringe_position())
```

```python
with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    initial_position = syringe.query_syringe_position()
    print("Position after power cycle:", initial_position)

    if initial_position != 1000:
        raise RuntimeError(
            f"Expected remembered position 1000 after power cycle, "
            f"got {initial_position}."
        )

    syringe.synchronize_syringe_position(wait=True)

    synced_position = syringe.query_syringe_position()
    print("Position after synchronization:", synced_position)

    if synced_position != 1000:
        raise RuntimeError(
            f"Expected position 1000 after synchronization, "
            f"got {synced_position}."
        )

    syringe.set_dynamic_speed_rpm(10)

    started = time.monotonic()
    syringe.move_plunger_to_position(1500, wait=True)
    elapsed = time.monotonic() - started

    final_position = syringe.query_syringe_position()
    print("Position after move:", final_position)
    print(f"Command-to-confirmation time: {elapsed:.2f} s")

    if final_position != 1500:
        raise RuntimeError(
            f"Expected position 1500 after move, got {final_position}."
        )
```

```python
# Pump power has remained on since the previous move.
# A new PC serial connection does not require re-synchronization.

with RunzeBus(port="COM5", trace=True) as bus:
    syringe = SY01B(bus, address=0x01)

    start_position = syringe.query_syringe_position()
    print("Initial position:", start_position)

    if start_position != 1500:
        raise RuntimeError(
            f"Expected position 1500 before this test, got {start_position}."
        )

    syringe.set_dynamic_speed_rpm(50)

    started = time.monotonic()
    syringe.move_plunger_to_position(5500, wait=True)
    elapsed = time.monotonic() - started

    final_position = syringe.query_syringe_position()
    print("Position after move:", final_position)
    print(f"Command-to-confirmation time: {elapsed:.2f} s")

    if final_position != 5500:
        raise RuntimeError(
            f"Expected position 5500 after move, got {final_position}."
        )
```
